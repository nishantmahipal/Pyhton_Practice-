```python
# write a program to display the salary slip details
empid = int(input("enter the employee id"))
emp_name = input("enter the name")
base_salary = float(input("enter the basic salary"))
da = base_salary*0.08 # daily allowance
ta = base_salary*0.12 # Travel allowance
bonus = int(input(" kitna bonus mila firr"))
pf = base_salary*0.15 #provident fund
tax = base_salary*0.05
net_salary = base_salary + da + ta + bonus
gross_salary = net_salary + pf +tax
```

```python
print(f"{"*"*10} Salary Slip {"*"*10}")
print ("EMPLOYEE ID",empid,"\nEMPLOYEE NAME",emp_name)
print (f" BASIC SALARY: {base_salary} \n DAILY ALLOWANCE IS: {da}\n Travel allowance is: {ta}")
print(f" your Bonus is : {bonus} \n your provident fund is: {pf} \n TAX dedection is: {tax}")
print(f" Net salary is: {net_salary} \n Gross Salary is: {gross_salary}")
print(f"{"*"*10} Thank you {"*"*10}")
```
